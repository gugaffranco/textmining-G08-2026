### **Search on how can i start implementing the project**

The pipeline should start with the **1. data collection and preprocessing**, where we can find in the first two notebook how to text clean the data! Some exemples are:
- Convert all text to lowercase
- Remove ponctuation, special characters and numbers
- Strip out common stopwords
- Aplly tokenization and lemmatiation. 

 
2. Feature Extraction & Vectorization

To turn words into numbers that models can interpret, use one of these approaches:
- Traditional Bag-of-Words / TF-IDF: Measures word frequency relative to how often it appears across all songs.
- Static Word Embeddings: Word2Vec or GloVe (e.g., Stanford GloVe vectors) to capture semantic meaning.
